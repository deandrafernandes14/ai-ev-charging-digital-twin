```python
import json
from pathlib import Path

data_path = Path("../Data/raw/acndata_sessions.json")

with open(data_path, "r", encoding="utf-8") as file:
    raw_data = json.load(file)

print(type(raw_data))
```

```python
print(raw_data.keys())
```

```python
sessions = raw_data["_items"]

print(type(sessions))
print("Number of sessions:", len(sessions))
```

```python
from pprint import pprint

pprint(sessions[0])
```

```python
print(sessions[0].keys())
```

```python
from collections import Counter

field_counts = Counter()

for session in sessions:
    field_counts.update(session.keys())

for field, count in field_counts.items():
    print(f"{field}: {count}")
```

```python
import pandas as pd

df = pd.DataFrame(sessions)

print("Dataset shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

missing_summary
```

```python
print("Fully duplicated rows:", df.duplicated().sum())
print("Duplicate session IDs:", df["sessionID"].duplicated().sum())
```

```python
print("Duplicate session IDs:", df["sessionID"].duplicated().sum())
print("Duplicate _id values:", df["_id"].duplicated().sum())
```

```python
print("Unique stations:", df["stationID"].nunique())
print("Unique spaces:", df["spaceID"].nunique())
print("Unique users:", df["userID"].nunique())
print("Total energy delivered:", round(df["kWhDelivered"].sum(), 2), "kWh")
```

```python
df["kWhDelivered"].describe()
```

```python
df["connectionTime"] = pd.to_datetime(df["connectionTime"], utc=True)
df["disconnectTime"] = pd.to_datetime(df["disconnectTime"], utc=True)
df["doneChargingTime"] = pd.to_datetime(df["doneChargingTime"], utc=True)

print("First connection:", df["connectionTime"].min())
print("Last connection:", df["connectionTime"].max())
```

```python
df["connectionTime_local"] = df["connectionTime"].dt.tz_convert("America/Los_Angeles")
df["disconnectTime_local"] = df["disconnectTime"].dt.tz_convert("America/Los_Angeles")
df["doneChargingTime_local"] = df["doneChargingTime"].dt.tz_convert("America/Los_Angeles")

print(df[["connectionTime", "connectionTime_local"]].head())
```

```python
df["connected_hours"] = (
    df["disconnectTime"] - df["connectionTime"]
).dt.total_seconds() / 3600

df["charging_hours"] = (
    df["doneChargingTime"] - df["connectionTime"]
).dt.total_seconds() / 3600

df[["kWhDelivered", "connected_hours", "charging_hours"]].describe()
```

```python
print("Negative charging durations:",
      (df["charging_hours"] < 0).sum())

print("Charging completion after disconnection:",
      (df["doneChargingTime"] > df["disconnectTime"]).sum())

print("Connected longer than 24 hours:",
      (df["connected_hours"] > 24).sum())

print("Charging longer than 24 hours:",
      (df["charging_hours"] > 24).sum())
```

```python
after_disconnect = df[df["doneChargingTime"] > df["disconnectTime"]].copy()

after_disconnect["difference_minutes"] = (
    after_disconnect["doneChargingTime"] - after_disconnect["disconnectTime"]
).dt.total_seconds() / 60

after_disconnect["difference_minutes"].describe()
```

```python
df.loc[
    df["charging_hours"] < 0,
    [
        "connectionTime",
        "doneChargingTime",
        "disconnectTime",
        "kWhDelivered",
        "stationID",
        "charging_hours"
    ]
]
```

```python
df.loc[
    df["connected_hours"] > 24,
    [
        "connectionTime",
        "disconnectTime",
        "kWhDelivered",
        "stationID",
        "connected_hours",
        "charging_hours"
    ]
].sort_values("connected_hours", ascending=False).head(10)
```
